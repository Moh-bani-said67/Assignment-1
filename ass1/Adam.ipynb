```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(100, 1) - 1
y = 2 * (X ** 2) + 3 * X + 5 + 0.2 * np.random.randn(100, 1)

X_poly = np.hstack((np.ones((100, 1)), X, X ** 2))

X_poly[:, 1:] = (X_poly[:, 1:] - np.mean(X_poly[:, 1:], axis=0)) / np.std(X_poly[:, 1:], axis=0)
def compute_cost(X, y, theta):
    m = len(y)
    predictions = X.dot(theta)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost
    # Adam Implementation
theta = np.random.randn(3, 1)
lr = 0.05
epochs = 200
beta1 = 0.9
beta2 = 0.999
epsilon = 1e-8

m_vec = np.zeros_like(theta)
v_vec = np.zeros_like(theta)
cost_history = []

m_len = len(y)

for t in range(1, epochs + 1):
    gradients = (1 / m_len) * X_poly.T.dot(X_poly.dot(theta) - y)

    m_vec = beta1 * m_vec + (1 - beta1) * gradients
    v_vec = beta2 * v_vec + (1 - beta2) * (gradients ** 2)

    m_hat = m_vec / (1 - beta1 ** t)
    v_hat = v_vec / (1 - beta2 ** t)

    theta = theta - (lr / (np.sqrt(v_hat) + epsilon)) * m_hat
    cost_history.append(compute_cost(X_poly, y, theta))

print("Final Theta (Adam):", theta.ravel())
```
