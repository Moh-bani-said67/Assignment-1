```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(100, 1) - 1
y = 2 * (X ** 2) + 3 * X + 5 + 0.2 * np.random.randn(100, 1)

X_poly = np.hstack((np.ones((100, 1)), X, X ** 2))

X_poly[:, 1:] = (X_poly[:, 1:] - np.mean(X_poly[:, 1:], axis=0)) / np.std(X_poly[:, 1:], axis=0)
def compute_cost(X, y, theta):
    m = len(y)
    predictions = X.dot(theta)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost
    # Initialization
theta = np.random.randn(3, 1)
lr = 0.03
epochs = 100
batch_size = 16
momentum = 0.9
velocity = np.zeros_like(theta)
cost_history = []

m = len(y)

for epoch in range(epochs):
    indices = np.random.permutation(m)
    X_shuffled = X_poly[indices]
    y_shuffled = y[indices]

    for i in range(0, m, batch_size):
        X_batch = X_shuffled[i:i+batch_size]
        y_batch = y_shuffled[i:i+batch_size]
        gradients = (1 / len(y_batch)) * X_batch.T.dot(X_batch.dot(theta) - y_batch)
        velocity = momentum * velocity + lr * gradients
        theta = theta - velocity

    cost_history.append(compute_cost(X_poly, y, theta))

print("Final Theta (Mini-Batch GD):", theta.ravel())
```
