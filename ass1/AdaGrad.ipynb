```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(100, 1) - 1
y = 2 * (X ** 2) + 3 * X + 5 + 0.2 * np.random.randn(100, 1)

X_poly = np.hstack((np.ones((100, 1)), X, X ** 2))

X_poly[:, 1:] = (X_poly[:, 1:] - np.mean(X_poly[:, 1:], axis=0)) / np.std(X_poly[:, 1:], axis=0)

def compute_cost(X, y, theta):
    m = len(y)
    predictions = X.dot(theta)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost
    # AdaGrad Implementation
theta = np.random.randn(3, 1)
lr = 0.1
epochs = 200
epsilon = 1e-8
G = np.zeros_like(theta)
cost_history = []
m = len(y)
for epoch in range(epochs):
    gradients = (1 / m) * X_poly.T.dot(X_poly.dot(theta) - y)
    G += gradients ** 2
    theta = theta - (lr / (np.sqrt(G) + epsilon)) * gradients
    cost_history.append(compute_cost(X_poly, y, theta))
print("Final Theta (AdaGrad):", theta.ravel())
```
